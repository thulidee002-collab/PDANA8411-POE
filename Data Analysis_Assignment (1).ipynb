```python
# Create boxplots to identify potential outliers

numerical_columns = ["age", "bmi", "children", "charges"]

for column in numerical_columns:

    plt.figure(figsize=(8, 4))

    sns.boxplot(x=df[column])

    plt.title(f"Boxplot of {column}")

    plt.xlabel(column)

    plt.show()
```

```python
# Calculate the number of potential outliers using the IQR method

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(f"{column}: {len(outliers)} potential outliers")
```

```python
# Create a copy of the dataset before preprocessing.
# Keeping the original dataframe unchanged allows us to return to the
# original data if necessary.

df_model = df.copy()

# Convert categorical variables into numerical dummy variables.
# drop_first=True removes one category from each categorical variable
# to avoid perfect multicollinearity.

df_model = pd.get_dummies(
    df_model,
    columns=["sex", "smoker", "region"],
    drop_first=True
)

# Display the first five rows after encoding
df_model.head()
```

```python
# Display the new column names after one-hot encoding

print("Columns after encoding:")
print(df_model.columns.tolist())
```

```python
# Check the data types after encoding

df_model.dtypes
```

```python
# Convert Boolean columns into numerical 0 and 1 values

boolean_columns = df_model.select_dtypes(include="bool").columns

df_model[boolean_columns] = df_model[boolean_columns].astype(int)

# Display the data types again
df_model.dtypes
```

```python
# Define the target variable

y = df_model["charges"]

# Define the predictor variables by removing the target column

X = df_model.drop("charges", axis=1)

# Display the dimensions of X and y

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# Display the predictor variables

X.head()
```

```python
# Import statsmodels for statistical modelling and P-value analysis

import statsmodels.api as sm
```

```python
# Add a constant/intercept to the predictor variables.
# statsmodels does not automatically add an intercept in this approach.

X_with_constant = sm.add_constant(X)

# Fit an Ordinary Least Squares regression model

ols_model = sm.OLS(y, X_with_constant).fit()

# Display the statistical summary

print(ols_model.summary())
```

```python
# Extract the P-values from the fitted OLS model

p_values = ols_model.pvalues

# Display the P-values

print(p_values)
```

```python
# Create a table containing the P-values

p_value_table = pd.DataFrame({
    "Feature": p_values.index,
    "P-value": p_values.values
})

p_value_table
```

```python
# Identify predictors with statistically significant P-values

significant_features = p_value_table[
    (p_value_table["Feature"] != "const") &
    (p_value_table["P-value"] < 0.05)
]

print("Statistically significant features:")
display(significant_features)
```

```python
# Import the Variance Inflation Factor function

from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
# Calculate VIF for each predictor variable

vif_data = pd.DataFrame()

vif_data["Feature"] = X_with_constant.columns

vif_data["VIF"] = [
    variance_inflation_factor(
        X_with_constant.values,
        i
    )
    for i in range(X_with_constant.shape[1])
]

vif_data
```

```python
# Import the train_test_split function

from sklearn.model_selection import train_test_split

# Split the dataset into training and testing data

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Display the sizes of the resulting datasets

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
# Import the Linear Regression algorithm
from sklearn.linear_model import LinearRegression

# Create the Linear Regression model
linear_model = LinearRegression()

# Train the model using the training data
linear_model.fit(X_train, y_train)

print("Linear Regression model trained successfully.")
```

```python
# Display the model intercept
print("Intercept:", linear_model.intercept_)

# Create a table containing each feature and its coefficient
coefficient_table = pd.DataFrame({
    "Feature": X_train.columns,
    "Coefficient": linear_model.coef_
})

# Display the coefficients
coefficient_table
```

```python
# Generate predictions using the test data
y_pred = linear_model.predict(X_test)

# Display the first ten predictions
print("First 10 predicted charges:")
print(y_pred[:10])
```

```python
# Create a comparison table of actual and predicted charges

comparison = pd.DataFrame({
    "Actual Charges": y_test.values,
    "Predicted Charges": y_pred
})

# Display the first 10 predictions
comparison.head(10)
```

```python
# Import regression evaluation metrics
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)
```

```python
# Calculate the R-squared score
r2 = r2_score(y_test, y_pred)

# Calculate Mean Absolute Error
mae = mean_absolute_error(y_test, y_pred)

# Calculate Mean Squared Error
mse = mean_squared_error(y_test, y_pred)

# Calculate Root Mean Squared Error
rmse = np.sqrt(mse)

# Display the results
print("Baseline Model Performance")
print("--------------------------")
print(f"R²   : {r2:.4f}")
print(f"MAE  : {mae:.2f}")
print(f"MSE  : {mse:.2f}")
print(f"RMSE : {rmse:.2f}")
```

**The R² of 0.7836 is reasonably strong for a first Linear Regression model, but there is still approximately 21.64% of the variation in charges that the model does not explain.**

**The MAE and RMSE also indicate that the model can make relatively substantial prediction errors for some clients.**

```python
# Generate predictions for the training data
y_train_pred = linear_model.predict(X_train)

# Calculate training R²
train_r2 = r2_score(y_train, y_train_pred)

# Calculate testing R²
test_r2 = r2_score(y_test, y_pred)

print("Training R²:", round(train_r2, 4))
print("Testing R² :", round(test_r2, 4))
```

```python
# Plot actual charges against predicted charges

plt.figure(figsize=(8, 6))

plt.scatter(y_test, y_pred, alpha=0.6)

# Create a reference line representing perfect predictions
minimum = min(y_test.min(), y_pred.min())
maximum = max(y_test.max(), y_pred.max())

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--"
)

plt.title("Actual vs Predicted Medical Charges")
plt.xlabel("Actual Charges")
plt.ylabel("Predicted Charges")

plt.show()
```

```python
# Calculate residuals
residuals = y_test - y_pred

# Display the first few residuals
residuals.head()
```

```python
# Plot residuals against predicted values

plt.figure(figsize=(9, 6))

plt.scatter(y_pred, residuals, alpha=0.6)

# Add a horizontal line at zero
plt.axhline(y=0, linestyle="--")

plt.title("Residual Plot")
plt.xlabel("Predicted Charges")
plt.ylabel("Residuals")

plt.show()
```

#### Interpretation of the Baseline Model

The baseline Linear Regression model achieved an R² score of 0.7836. This means that approximately 78.36% of the variation in medical insurance charges in the testing dataset can be explained by the predictor variables used in the model.

The model produced a Mean Absolute Error (MAE) of approximately 4,181.19. This means that the model's predictions differ from the actual medical charges by approximately 4,181.19 charge units on average.

The Root Mean Squared Error (RMSE) was approximately 5,796.28. Since RMSE gives greater weight to large errors, the difference between the MAE and RMSE suggests that some observations have relatively large prediction errors.

The baseline model therefore provides a reasonable starting point, but there is room for improvement. The model does not explain approximately 21.64% of the variation in medical charges, and the prediction errors may be significant for some clients.


```python
# Create a copy of the encoded dataset for the improved model

df_improved = df_model.copy()

# Create an interaction between age and smoking status
df_improved["age_smoker"] = (
    df_improved["age"] * df_improved["smoker_yes"]
)

# Create an interaction between BMI and smoking status
df_improved["bmi_smoker"] = (
    df_improved["bmi"] * df_improved["smoker_yes"]
)

# Display the new features
df_improved[[
    "age",
    "bmi",
    "smoker_yes",
    "age_smoker",
    "bmi_smoker"
]].head()
```

```python
# Separate the target variable

y_improved = df_improved["charges"]

# Separate the predictor variables

X_improved = df_improved.drop("charges", axis=1)

# Display the features
X_improved.head()
```

```python
# Split the improved dataset into training and testing data

X_train_improved, X_test_improved, y_train_improved, y_test_improved = train_test_split(
    X_improved,
    y_improved,
    test_size=0.20,
    random_state=42
)

print("Improved training data:", X_train_improved.shape)
print("Improved testing data:", X_test_improved.shape)
```

```python
# Create a new Linear Regression model

improved_model = LinearRegression()

# Train the model using the engineered features

improved_model.fit(
    X_train_improved,
    y_train_improved
)

print("Improved Linear Regression model trained successfully.")
```

```python
# Generate predictions using the improved model

y_pred_improved = improved_model.predict(
    X_test_improved
)

# Display the first 10 predictions

print("First 10 improved predictions:")
print(y_pred_improved[:10])
```

```python
# Calculate improved model performance metrics

improved_r2 = r2_score(
    y_test_improved,
    y_pred_improved
)

improved_mae = mean_absolute_error(
    y_test_improved,
    y_pred_improved
)

improved_mse = mean_squared_error(
    y_test_improved,
    y_pred_improved
)

improved_rmse = np.sqrt(improved_mse)

# Display the results

print("Improved Model Performance")
print("--------------------------")
print(f"R²   : {improved_r2:.4f}")
print(f"MAE  : {improved_mae:.2f}")
print(f"MSE  : {improved_mse:.2f}")
print(f"RMSE : {improved_rmse:.2f}")
```

```python
# Create a comparison table between the baseline and improved models

model_comparison = pd.DataFrame({
    "Metric": ["R²", "MAE", "MSE", "RMSE"],
    "Baseline Model": [
        r2,
        mae,
        mse,
        rmse
    ],
    "Improved Model": [
        improved_r2,
        improved_mae,
        improved_mse,
        improved_rmse
    ]
})

model_comparison
```

#### Retraining and Model Improvement

The baseline model achieved an R² score of 0.7836, with an MAE of 4,181.19 and an RMSE of 5,796.28.

Although the baseline model provided a reasonable level of predictive performance, the EDA indicated that smoking status has a particularly strong relationship with medical charges. It was therefore considered possible that the relationship between age/BMI and charges could differ between smokers and non-smokers.

To investigate this, two interaction features were introduced:

- `age_smoker`
- `bmi_smoker`

The model was retrained using these additional features while keeping the same 80/20 train-test split and random state. This allowed the performance of the improved model to be compared fairly with the baseline model.

The improved model was evaluated using the same metrics: R², MAE, MSE, and RMSE.

The model will be considered an improvement if it produces a higher R² score and lower MAE and RMSE compared with the baseline model.

```python
# Calculate the percentage improvement in the evaluation metrics

r2_percentage_change = ((improved_r2 - r2) / r2) * 100

mae_percentage_reduction = ((mae - improved_mae) / mae) * 100

mse_percentage_reduction = ((mse - improved_mse) / mse) * 100

rmse_percentage_reduction = ((rmse - improved_rmse) / rmse) * 100

print(f"R² percentage increase: {r2_percentage_change:.2f}%")
print(f"MAE percentage reduction: {mae_percentage_reduction:.2f}%")
print(f"MSE percentage reduction: {mse_percentage_reduction:.2f}%")
print(f"RMSE percentage reduction: {rmse_percentage_reduction:.2f}%")
```

#### Conclusion

This study developed a Linear Regression model to estimate medical insurance charges using age, sex, BMI, number of children, smoking status, and geographical region.

An initial Exploratory Data Analysis was conducted to understand the structure, quality, distributions, relationships, and potential outliers within the dataset.

The categorical variables were converted into numerical representations using one-hot encoding. The dataset was then divided into training and testing subsets using an 80/20 split.

A baseline Linear Regression model was initially trained. It achieved an R² score of 0.7836, an MAE of 4,181.19, an MSE of 33,596,915.85, and an RMSE of 5,796.28.

Feature engineering was subsequently used to create interaction features between smoking status and age and between smoking status and BMI. The Linear Regression model was retrained using these additional features.

The improved model achieved an R² score of 0.8652, an MAE of 2,757.76, an MSE of 20,922,607.94, and an RMSE of 4,574.12. Therefore, the improved model performed better than the baseline model across all four evaluation metrics.

The results demonstrate that the interaction between smoking status and other client characteristics can provide useful information when estimating medical insurance charges.

The final model is suitable as a proof-of-concept model. However, the dataset is US-based and should not be directly used to determine South African medical aid pricing. Before deployment, the model should be retrained and validated using representative South African medical aid data and additional relevant client and healthcare variables.
